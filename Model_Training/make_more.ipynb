```python
g = torch.Generator().manual_seed(2147483647)
ix = 0
out = []
while True:
  # If the sum of counts for the current character is zero, it means
  # this character never precedes any other in the training data.
  # This is especially true for index 27 ('<E>') as it's not meant to be followed by anything.
  # In such a case, we should break the loop to prevent division by zero.
  current_row_sum = T[ix].sum().float()
  if current_row_sum == 0:
    break

  p = T[ix].float()
  p = p / current_row_sum
  ix = torch.multinomial(p,num_samples=1, replacement=True, generator = g).item()
  out.append(itos[ix])
  # The loop already breaks if the end token '.' (index 0) is generated.
  if ix == 0:
    break
print(''.join(out))
```

```python
#context_length = 1 character
#vocab_size = 27 characters

#output is not that great


g = torch.Generator().manual_seed(2147483647)

for i in range(50):

  ix = 0
  out = []

  while True:
    p = T[ix].float()
    p = p / p.sum()
    ix = torch.multinomial(p,num_samples=1, replacement=True, generator = g).item()
    out.append(itos[ix])
    if ix == 0:
      break
  print(''.join(out))
```

```python
##### Neural_Network_Approach ######

#prvious approach is not recommended..as the context size goes up...the table size grows very exponentially
# The same can be done by using Neural_Network_Approach

words = open('names.txt','r').read().splitlines()
```

```python
#Neural network approach

xs ,ys = [],[]


for w in words[0:1]:
  chs = ['.'] + list(w) + ['.']
  for ch1,ch2 in zip(chs,chs[1:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    xs.append(ix1)
    ys.append(ix2)


xs = torch.tensor(xs)
ys = torch.tensor(ys)
print(xs)
print(ys)
```

```python
# now we have to think how to feed in these examples to a neural net
# we can encode these inputs usin g one hot encoding

import torch.nn.functional as F
xenc = F.one_hot(xs,num_classes=27).float()
xenc
```
